## Հուսալիության ընդլայնում
Գիտական NLI-ի 659 ուսուցման և 150 կարգավորման պնդումներից հետո պաշտոնական SciFact dev-ի 300 պնդումների նախագծային վերջնական ստուգումը տվել է 56.33% accuracy և 54.07% macro-F1։ Սա անգլերեն գիտական պնդումների փորձնական ռեժիմ է։
FakeNewsNet զգայուն ռեժիմը fake recall-ը բարձրացնում է 64.18%-ից 77.92%, իսկ precision-ը նվազեցնում է 73.10%-ից 60.64%։ Ամբողջ չափումները պահպանված են reports/reliability պանակում։
